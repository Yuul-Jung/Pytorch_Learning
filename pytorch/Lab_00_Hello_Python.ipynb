## 실습 단계

### 1. 실행환경 확인


```python
import platform
import sys

print("Python 버전:", sys.version.split()[0])
print("운영체제:", platform.system())
print("첫 코드가 정상적으로 실행되었습니다.")
```

### 2. 숫자와 산술연산

Python에서는 사칙연산과 거듭제곱을 바로 계산할 수 있습니다.


```python
print("덧셈:", 10 + 3)
print("뺄셈:", 10 - 3)
print("곱셈:", 10 * 3)
print("나눗셈:", 10 / 3)
print("몫:", 10 // 3)
print("나머지:", 10 % 3)
print("거듭제곱:", 10 ** 2)
```

### 3. 변수와 자료형


```python
temperature = 31.5
sensor_name = "온도 센서"
is_running = True

print(temperature, type(temperature))
print(sensor_name, type(sensor_name))
print(is_running, type(is_running))
```

### 4. 리스트와 인덱싱

리스트는 여러 값을 순서대로 저장합니다. 인덱스는 `0`부터 시작합니다.


```python
sensor_values = [20.1, 21.3, 22.8, 24.0]

print("전체 값:", sensor_values)
print("첫 번째 값:", sensor_values[0])
print("마지막 값:", sensor_values[-1])
print("앞의 두 값:", sensor_values[:2])
print("데이터 개수:", len(sensor_values))
```

### 5. 조건문과 함수

함수는 입력을 받아 정해진 작업을 수행하고 결과를 반환합니다.


```python
def classify_temperature(value):
    # 설비 상태 판정
    if value < 25.0:
        return "정상"
    elif value < 35.0:
        return "주의"
    else:
        return "고온"

state = classify_temperature(temperature)
print(f"{temperature}℃의 상태: {state}")
```

### 6. 반복문으로 센서값 처리

반복문은 여러 데이터를 같은 방법으로 처리합니다.


```python
total = 0.0

for index, value in enumerate(sensor_values):
    state = classify_temperature(value)
    print(f"{index + 1}번째 값: {value:.1f}℃ → {state}")
    total += value

average = total / len(sensor_values)
print(f"평균 온도: {average:.2f}℃")
```
