# NumPy와 PyTorch 환경 점검

이 노트북은 Ubuntu 가상환경에서 NumPy와 PyTorch가 제대로 설치되어 있는지 빠르게 확인합니다.

먼저 현재 Python 실행 경로와 버전을 확인하고, 필요한 패키지 임포트를 시도합니다.


````python
import sys
print("Python executable:", sys.executable)
print("Python version:", sys.version)
````

## 패키지 버전 확인

다음 셀은 NumPy와 PyTorch를 import하고 버전을 출력합니다. 가상환경에서 실행 중인지 확인할 수 있습니다.


````python
import numpy as np
print("NumPy version:", np.__version__)

import torch
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
````

## 문제 해결

- `ModuleNotFoundError: No module named "torch"`가 발생하면, 가상환경이 활성화된 상태에서 `python -m pip install torch`를 실행하세요.
- `sys.executable` 경로가 가상환경 경로가 아니라면, 터미널에서 `source ~/python_envs/pytorch_env/bin/activate` 후 다시 실행하세요.
