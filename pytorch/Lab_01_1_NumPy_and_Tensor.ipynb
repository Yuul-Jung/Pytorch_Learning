# Lab 01_1. NumPy와 Tensor

## 목표

- Python(파이썬) 리스트와 NumPy(넘파이) 배열 비교하기
- Scalar(스칼라, 값 1개)·Vector(벡터, 1차원 배열)·Matrix(매트릭스, 행렬) 구분하기
- NumPy 배열의 산술연산·브로드캐스팅·원소 접근 익히기
- PyTorch Tensor(파이토치 텐서, 다차원 배열) 만들기
- Tensor의 `dim`, `shape`, `dtype`, `device` 확인하기


## 실습 단계

### 1. NumPy와 PyTorch 불러오기


```python
import numpy as np
import torch

print("NumPy 버전:", np.__version__)
print("PyTorch 버전:", torch.__version__)
print("GPU 사용 가능:", torch.cuda.is_available())
```

### 2. Python 리스트와 NumPy 배열 비교

Python 리스트의 `+`는 두 리스트를 이어 붙입니다. NumPy 배열의 `+`는 같은 위치의 원소끼리 더합니다.

> **A NumPy array performs numerical operations element by element.**  
> **NumPy 배열은 같은 위치의 원소별로 수치 연산을 수행합니다.**


```python
python_list_a = [1.0, 2.0, 3.0]
python_list_b = [2.0, 4.0, 6.0]

numpy_array_a = np.array([1.0, 2.0, 3.0])
numpy_array_b = np.array([2.0, 4.0, 6.0])

print("Python 리스트 + :", python_list_a + python_list_b)
print("NumPy 배열 + :", numpy_array_a + numpy_array_b)
print("배열 자료형:", type(numpy_array_a))
```

### 3. NumPy 배열의 산술연산

Element-wise operation(엘리먼트 와이즈 오퍼레이션, 원소별 연산)은 같은 위치의 원소끼리 계산합니다.


```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])

print("덧셈:", x + y)
print("뺄셈:", x - y)
print("원소별 곱셈:", x * y)
print("원소별 나눗셈:", x / y)
print("스칼라 나눗셈:", x / 2.0)
```

### 4. Scalar·Vector·Matrix 구분

- Scalar(스칼라): 값 1개, 형상 `()`
- Vector(벡터): 1차원 배열, 형상 `(3,)`
- Matrix(매트릭스): 2차원 행렬, 형상 `(2, 2)`
- Tensor(텐서): 위 구조를 포함하는 다차원 배열, 형상 `(2, 3, 4)` 등

`ndim`은 차원의 개수이고, `shape`는 각 차원의 크기입니다.


```python
scalar_np = np.array(7)
vector_np = np.array([1, 2, 3])
matrix_np = np.array([[1, 2], [3, 4]])

for name, value in [
    ("Scalar", scalar_np),
    ("Vector", vector_np),
    ("Matrix", matrix_np),
]:
    print(f"{name:6} → ndim: {value.ndim}, shape: {value.shape}")
```

### 5. NumPy N차원 배열과 속성

2차원 배열에서 바깥쪽 대괄호는 행의 묶음이고, 안쪽 대괄호 하나가 한 행입니다.


```python
A = np.array([[1, 2], [3, 4]])

print(A)
print("차원 수:", A.ndim)
print("형상:", A.shape)
print("원소 자료형:", A.dtype)
```

### 6. 배열 연산과 Broadcasting

Broadcasting(브로드캐스팅, 작은 배열을 큰 배열의 형상에 맞춰 자동 확장하는 기능)을 사용하면 형상이 다른 배열도 일정한 규칙으로 계산할 수 있습니다.


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[3, 0], [0, 6]])
row_scale = np.array([10, 20])

print("A + B")
print(A + B)

print("\nA * B: 원소별 곱셈")
print(A * B)

print("\nA * 10: 스칼라 브로드캐스팅")
print(A * 10)

print("\nA * [10, 20]: 행 방향 브로드캐스팅")
print(A * row_scale)
```

### 7. 원소 접근·평탄화·조건 선택

- Indexing(인덱싱, 위치로 원소 선택)
- Flatten(플래튼, 다차원 배열을 1차원으로 평탄화)
- Boolean Mask(불리언 마스크, 조건이 참인 원소 선택)


```python
X = np.array([[51, 55], [14, 19], [0, 4]])

print("전체 배열")
print(X)
print("첫 번째 행:", X[0])
print("첫 번째 행의 두 번째 값:", X[0][1])

X_flat = X.flatten()
print("평탄화:", X_flat)
print("0·2·4번 원소:", X_flat[np.array([0, 2, 4])])
print("15 초과 여부:", X_flat > 15)
print("15 초과 값:", X_flat[X_flat > 15])
```

### 8. PyTorch Tensor 만들기

Tensor(텐서)는 Scalar·Vector·Matrix와 그 이상의 다차원 배열을 모두 포함하는 표현입니다.

> **A tensor is a multidimensional array that can run on a CPU or GPU.**  
> **텐서는 CPU(씨피유) 또는 GPU(지피유)에서 연산할 수 있는 다차원 배열입니다.**


```python
scalar = torch.tensor(7)
vector = torch.tensor([1.0, 2.0, 3.0])
matrix = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:")
print(matrix)
```

### 9. Tensor의 dim·shape·dtype·device

- `dim()`(딤): 차원의 개수
- `shape`(셰이프): 각 차원의 크기
- `dtype`(디타입): 원소의 자료형
- `device`(디바이스): 텐서가 저장된 CPU 또는 GPU


```python
sensor_tensor = torch.tensor(
    [[22.5, 0.10], [31.0, 0.45], [36.0, 0.90]],
    dtype=torch.float32,
)

print(sensor_tensor)
print("dim:", sensor_tensor.dim())
print("shape:", sensor_tensor.shape)
print("dtype:", sensor_tensor.dtype)
print("device:", sensor_tensor.device)
```

### 10. NumPy 배열과 Tensor 변환

NumPy 배열은 `torch.from_numpy()`로 Tensor로 바꿀 수 있고, CPU Tensor는 `.numpy()`로 NumPy 배열로 바꿀 수 있습니다.


```python
numpy_data = np.array([[1.0, 2.0], [3.0, 4.0]], dtype=np.float32)
tensor_data = torch.from_numpy(numpy_data)
numpy_again = tensor_data.numpy()

print("NumPy → Tensor")
print(tensor_data)
print("Tensor → NumPy")
print(numpy_again)
```

### 11. Tensor의 device 선택

GPU가 있으면 CUDA(쿠다, 엔비디아 GPU 병렬 연산 플랫폼)를 선택하고, 없으면 CPU를 사용합니다. Colab의 런타임에서 GPU를 선택하면 같은 코드가 자동으로 GPU를 사용합니다.


```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
tensor_on_device = sensor_tensor.to(device)

print("선택된 device:", device)
print("Tensor의 실제 device:", tensor_on_device.device)
```
