이 노트북은 앞에서 배운 NumPy(넘파이, 수치 배열 라이브러리) 배열을  
PyTorch Tensor(파이토치 텐서, 다차원 수치 배열)로 연결하고 기본 연산을 실습합니다.

## Goal(골, 학습 목표)

- Python List(파이썬 리스트)와 NumPy 배열로 Tensor 만들기
- `dim`, `shape`, `dtype`, `device` 확인하기
- Tensor의 인덱싱과 슬라이싱 사용하기
- 원소별 곱과 행렬곱 구분하기
- `reshape`, `cat`, `stack`으로 Tensor 모양 다루기


## Setup

Colab에는 PyTorch가 기본 설치되어 있습니다.  
아래 셀을 실행해 버전과 GPU(지피유, 그래픽 연산장치) 사용 가능 여부를 확인합니다.


```python
import numpy as np
import torch

print("PyTorch 버전:", torch.__version__)
print("GPU 사용 가능:", torch.cuda.is_available())
```

## Steps

### 1. Python 리스트로 Tensor 만들기

`torch.tensor()`는 입력 데이터의 자료형을 기준으로 Tensor를 만듭니다.  
딥러닝 계산에는 주로 `float32`를 사용하므로 `dtype`을 명시해 봅니다.


```python
list_data = [[10, 20], [30, 40]]

tensor_from_list = torch.tensor(list_data, dtype=torch.float32)

print(tensor_from_list)
print("dim:", tensor_from_list.dim())
print("shape:", tensor_from_list.shape)
print("dtype:", tensor_from_list.dtype)
print("device:", tensor_from_list.device)
```

### 2. NumPy 배열과 Tensor 변환

- `torch.from_numpy()` : NumPy 배열을 Tensor로 변환
- `.numpy()` : CPU Tensor를 NumPy 배열로 변환

`torch.from_numpy()`로 만든 Tensor와 원본 NumPy 배열은 메모리를 공유할 수 있습니다.  
한쪽 값을 바꾸면 다른 쪽 값도 바뀔 수 있다는 점에 주의합니다.


```python
numpy_data = np.array([[10, 20], [30, 40]], dtype=np.float32)
tensor_from_numpy = torch.from_numpy(numpy_data)

print("NumPy 배열:")
print(numpy_data)
print("\nTensor:")
print(tensor_from_numpy)
print("dtype:", tensor_from_numpy.dtype)
```

```python
# 메모리 공유 확인
numpy_data[0, 0] = 99

print("변경된 NumPy 배열:")
print(numpy_data)
print("\n같은 메모리를 공유하는 Tensor:")
print(tensor_from_numpy)

# 독립된 복사본이 필요하면 clone() 사용
tensor_copy = tensor_from_numpy.clone()
print("\n독립된 Tensor 복사본:")
print(tensor_copy)
```

### 3. CPU와 GPU 장치 선택

`device`는 Tensor가 저장되고 계산되는 위치입니다.

- `cpu` : CPU(씨피유, 중앙처리장치) 메모리
- `cuda` : NVIDIA GPU 메모리

GPU가 없는 환경에서도 오류가 나지 않도록 자동 선택합니다.


```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
tensor_on_device = tensor_from_list.to(device)

print("선택된 장치:", device)
print("Tensor 장치:", tensor_on_device.device)
```

### 4. 일정한 값과 난수로 Tensor 만들기

- `zeros` : 모든 원소가 0
- `ones` : 모든 원소가 1
- `arange` : 일정한 간격의 연속 값
- `rand` : 0 이상 1 미만의 균등분포 난수
- `randn` : 평균 0, 표준편차 1인 정규분포 난수

`manual_seed()`를 사용하면 다시 실행해도 같은 난수가 생성되어 결과를 비교하기 쉽습니다.


```python
torch.manual_seed(42)

zeros_tensor = torch.zeros(2, 3)
ones_tensor = torch.ones(2, 3)
range_tensor = torch.arange(0, 6).reshape(2, 3)
uniform_tensor = torch.rand(2, 3)
normal_tensor = torch.randn(2, 3)

print("zeros:\n", zeros_tensor)
print("ones:\n", ones_tensor)
print("arange + reshape:\n", range_tensor)
print("rand:\n", uniform_tensor)
print("randn:\n", normal_tensor)
```

### 5. 인덱싱과 슬라이싱

Tensor도 Python 리스트와 NumPy 배열처럼 위치를 지정해 값을 선택합니다.

- `tensor[0]` : 0번 행
- `tensor[:, 1:]` : 모든 행에서 1번 열 이후
- `tensor[-1, -1]` : 마지막 행의 마지막 원소


```python
tensor_a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.float32
)

print("전체 Tensor:\n", tensor_a)
print("0번 행:", tensor_a[0])
print("모든 행의 1번 열 이후:\n", tensor_a[:, 1:])
print("마지막 원소:", tensor_a[-1, -1])
print("조건 선택:", tensor_a[tensor_a > 3])
```

### 6. 기본 산술 연산과 원소별 곱

Element-wise Operation(엘리먼트 와이즈 오퍼레이션, 원소별 연산)은  
같은 위치에 있는 원소끼리 계산합니다.

`tensor_a * tensor_b`와 `torch.mul(tensor_a, tensor_b)`는 같은 원소별 곱입니다.


```python
tensor_b = torch.tensor(
    [[7, 8, 9],
     [10, 11, 12]],
    dtype=torch.float32
)

print("덧셈:\n", tensor_a + tensor_b)
print("뺄셈:\n", tensor_a - tensor_b)
print("원소별 곱:\n", tensor_a * tensor_b)
print("원소별 나눗셈:\n", tensor_b / tensor_a)
```

### 7. 원소별 곱과 행렬곱 구분

| 연산 | 코드 | 계산 기준 |
|---|---|---|
| 원소별 곱 | `A * B` | 같은 위치의 원소끼리 곱함 |
| Matrix Multiplication(매트릭스 멀티플리케이션, 행렬곱) | `A @ B` | A의 열 수와 B의 행 수가 같아야 함 |

`tensor_a`와 `tensor_b`의 shape가 모두 `(2, 3)`이므로 바로 행렬곱할 수 없습니다.  
두 번째 Tensor를 `(3, 2)`로 바꾸면 `(2, 3) @ (3, 2)`가 되어 행렬곱이 가능합니다.


```python
print("tensor_a shape:", tensor_a.shape)
print("tensor_b shape:", tensor_b.shape)

if tensor_a.shape[1] != tensor_b.shape[0]:
    print("바로 행렬곱할 수 없음: A의 열 수와 B의 행 수가 다릅니다.")

tensor_b_reshaped = tensor_b.reshape(3, 2)
matrix_product = tensor_a @ tensor_b_reshaped

print("\n변경된 tensor_b shape:", tensor_b_reshaped.shape)
print("행렬곱 결과 shape:", matrix_product.shape)
print("행렬곱 결과:\n", matrix_product)
```

### 8. Tensor 형상 변경

`reshape()`는 원소의 총개수를 유지하면서 Tensor의 모양을 바꿉니다.

`(2, 3)` Tensor의 원소는 6개이므로 `(3, 2)`, `(1, 6)`, `(6,)` 등으로 바꿀 수 있습니다.


```python
print("원래 shape:", tensor_a.shape)
print("reshape(3, 2):\n", tensor_a.reshape(3, 2))
print("reshape(1, 6):\n", tensor_a.reshape(1, 6))
print("flatten():", tensor_a.flatten())
```

### 9. Tensor 합치기

- `torch.cat()` : 기존 차원을 따라 연결
- `torch.stack()` : 새로운 차원을 만들어 쌓기

`dim=0`과 `dim=1`은 단순히 행·열이라는 이름보다  
**어느 차원의 크기를 늘리는가**로 이해하는 것이 정확합니다.


```python
cat_dim0 = torch.cat([tensor_a, tensor_b], dim=0)
cat_dim1 = torch.cat([tensor_a, tensor_b], dim=1)
stack_dim0 = torch.stack([tensor_a, tensor_b], dim=0)

print("cat dim=0 shape:", cat_dim0.shape)
print(cat_dim0)

print("\ncat dim=1 shape:", cat_dim1.shape)
print(cat_dim1)

print("\nstack dim=0 shape:", stack_dim0.shape)
print(stack_dim0)
```

## Practice

온도와 진동 센서 데이터를 Tensor로 만들고 다음을 확인합니다.

1. `dim`, `shape`, `dtype`, `device`
2. 첫 번째 샘플
3. 온도 열 전체
4. 온도가 30 이상인 샘플
5. 각 센서 값에 보정계수를 원소별로 곱한 결과


```python
# 각 행은 [온도, 진동] 한 샘플입니다.
sensor_data = torch.tensor(
    [[22.0, 0.10],
     [25.0, 0.15],
     [31.0, 0.45],
     [36.0, 0.90]],
    dtype=torch.float32
)

correction = torch.tensor([1.0, 10.0])
corrected_sensor_data = sensor_data * correction

print("dim:", sensor_data.dim())
print("shape:", sensor_data.shape)
print("dtype:", sensor_data.dtype)
print("device:", sensor_data.device)
print("\n첫 번째 샘플:", sensor_data[0])
print("온도 열:", sensor_data[:, 0])
print("온도 30 이상인 샘플:\n", sensor_data[sensor_data[:, 0] >= 30])
print("보정 결과:\n", corrected_sensor_data)
```

아래 셀의 오류가 없으면 핵심 결과가 올바릅니다.

## 핵심 정리

- Tensor는 Python 리스트와 NumPy 배열로 만들 수 있습니다.
- `dim`은 차원의 개수, `shape`는 각 차원의 크기입니다.
- `dtype`은 원소의 자료형, `device`는 저장·연산 위치입니다.
- `*`는 원소별 곱이고 `@`는 행렬곱입니다.
- 행렬곱 전에는 두 Tensor의 shape가 맞는지 확인해야 합니다.
- `cat`은 기존 차원으로 연결하고 `stack`은 새 차원을 만듭니다.
